```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('Air_Quality.csv')
```

```python
#1.Data Cleaning
```

```python
# Handling Missing Values
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df=df.drop_duplicates()
df
```

```python
df = df.drop(columns=['Message'])
display(df.head())
```

```python
df["Measure Info"]=df["Measure Info"].str.capitalize()
df
```

```python
df["Name"] = df["Name"].str.strip()
df
```

```python
df["Geo Type Name"]=df["Geo Type Name"].str.strip()
df
```

```python
df.isnull().sum()
```

```python
df
```

```python
df.isnull().sum()
```

```python
df["Measure"].fillna("Mean", inplace=True)
df
```

```python
df["Measure Info"].unique()
```

```python
df["Geo Type Name"].unique()
```

```python
# 2.Data Preprocessing
```

```python
# Outlier Removing & Handling
```

```python
plt.boxplot(df["Data Value"])
plt.xlabel="Data"
plt.title="Boxplot"
plt.show()
```

```python
df["Data Value"].describe()
```

```python
Q1=df["Data Value"].quantile(0.25)
Q3=df["Data Value"].quantile(0.75)
IQR=Q3-Q1
lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR
df_cleaned=df[(df["Data Value"]>=lower_bound)&(df["Data Value"]<=upper_bound)]
df_cleaned
```

```python
# Outlier Handling Using Clip
```

```python
df["Data Value"]=df["Data Value"].clip(lower=lower_bound,upper=upper_bound)
df
```

```python
# Min-Max Normalization
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
df["Data Value"]=scaler.fit_transform(df[["Data Value"]])
df
```

```python
# Label Encoding
```

```python
from sklearn.preprocessing import LabelEncoder
label_encoder=LabelEncoder()
df['Measure Info']= label_encoder.fit_transform(df['Measure Info'])
df
```

```python
# One Hot Encoding
```

```python
from sklearn.preprocessing import OneHotEncoder
one_hot_encoder=OneHotEncoder(sparse_output=False)
encoded_data=one_hot_encoder.fit_transform(df[['Geo Type Name']])
encoded_data
```

```python

data_encoded=pd.DataFrame(encoded_data,columns=one_hot_encoder.get_feature_names_out(['Geo Type Name']))
data_encoded
```

```python
final_df=pd.concat([df,data_encoded],axis=1)
final_df
```

### 3. Data Visualization for Analysis

```python
# Bar chart for the distribution of 'Name' (Air Quality Indicators)
plt.figure(figsize=(12, 6))
df['Name'].value_counts().plot(kind='bar')
plt.title('Distribution of Air Quality Indicators')
plt.xlabel('Indicator Name')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Convert 'Start_Date' to datetime objects for time series analysis
df['Start_Date'] = pd.to_datetime(df['Start_Date'])

# Sort by date for proper time series plotting
df_sorted = df.sort_values(by='Start_Date')

# Line plot of 'Data Value' over 'Start_Date'
plt.figure(figsize=(14, 7))
plt.plot(df_sorted['Start_Date'], df_sorted['Data Value'])
plt.title('Data Value Trend Over Time')
plt.xlabel('Start Date')
plt.ylabel('Data Value')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Bar chart of average 'Data Value' by 'Geo Type Name'
average_data_value_by_geo = df.groupby('Geo Type Name')['Data Value'].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
average_data_value_by_geo.plot(kind='bar', color='skyblue')
plt.title('Average Data Value by Geo Type Name')
plt.xlabel('Geo Type Name')
plt.ylabel('Average Data Value')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
