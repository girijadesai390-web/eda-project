```python
#Outlier Remove (3 cols)
```

```python
Q1 = df["transaction_qty"].quantile(0.25)
Q3 = df["transaction_qty"].quantile(0.75)
IQR = Q3 - Q1

IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df = df[(df["transaction_qty"] >= lower_bound) & (df["transaction_qty"] <= upper_bound)]
df
```

```python
Q1 = df["Revenue"].quantile(0.25)
Q3 = df["Revenue"].quantile(0.75)
IQR = Q3 - Q1

IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df = df[(df["Revenue"] >= lower_bound) & (df["Revenue"] <= upper_bound)]
df
```

```python
Q1 = df["unit_price"].quantile(0.25)
Q3 = df["unit_price"].quantile(0.75)
IQR = Q3 - Q1

IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df = df[(df["unit_price"] >= lower_bound) & (df["unit_price"] <= upper_bound)]
df
```
